```python
import pandas as pd

# Load the data
df = pd.read_csv("churn.csv")

# 1. Basic shape and structure
print("Shape:", df.shape)
print("\nData types:\n", df.dtypes)
print("\nFirst few rows:\n", df.head())

# 2. Missing values
print("\nMissing values per column:\n", df.isnull().sum())

# 3. Duplicates
print("\nFully duplicated rows:", df.duplicated().sum())
print("Duplicate CustomerIds:", df['CustomerId'].duplicated().sum())

# 4. Obvious errors / range checks on key numeric columns
print("\nSummary stats:\n", df.describe())

# Specific sanity checks
print("\nAge range:", df['Age'].min(), "-", df['Age'].max())
print("CreditScore range:", df['CreditScore'].min(), "-", df['CreditScore'].max())
print("Balance range:", df['Balance'].min(), "-", df['Balance'].max())
print("Tenure range:", df['Tenure'].min(), "-", df['Tenure'].max())

# 5. Check categorical columns for unexpected/inconsistent values
for col in ['Geography', 'Gender', 'Card Type']:
    print(f"\nUnique values in {col}:", df[col].unique())

# 6. Check binary/flag columns only contain 0/1
for col in ['HasCrCard', 'IsActiveMember', 'Exited', 'Complain']:
    print(f"\nUnique values in {col}:", df[col].unique())
```
